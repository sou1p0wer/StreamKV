```python
# %%
import os
import shutil

# 路径定义
videos_dir = '/home/cyl476530/ReKV/data/StreamingBench/src/data/videos'
repaired_dir = '/home/cyl476530/ReKV/data/StreamingBench/src/data/repaired'

# 获取文件名集合
video_files = set(os.listdir(videos_dir))
repaired_files = set(os.listdir(repaired_dir))

# 需要拷贝的文件集合
to_copy = video_files - repaired_files

print(f"视频目录总数：{len(video_files)}")
print(f"repaired目录已有数：{len(repaired_files)}")
print(f"需拷贝文件数：{len(to_copy)}")
print(f"需拷贝文件：{list(to_copy)[:10]}{'...' if len(to_copy)>10 else ''}")

```

```python
for fname in to_copy:
    src = os.path.join(videos_dir, fname)
    dst = os.path.join(repaired_dir, fname)
    try:
        shutil.copy2(src, dst)
        print(f"Copied: {fname}")
    except Exception as e:
        print(f"Failed to copy {fname}: {e}")

print("完成！")
```
