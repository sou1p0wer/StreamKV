```python
# %%
import json

# 读取原始json文件（如果你用文件，若是用内存变量，则把data = ... 赋值进去即可）
with open('/home/cyl476530/ReKV/data/StreamingBench/src/data/question_real_rekv_online.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

# 遍历所有conversations条目，修改end_time
for item in data:
    for conv in item.get('conversations', []):
        if 'end_time' in conv and conv['end_time'] <= 1:
            conv['end_time'] = 2

# 可选：展示前2条数据
print(json.dumps(data[:2], ensure_ascii=False, indent=2))

# 保存结果
with open('/home/cyl476530/ReKV/data/StreamingBench/src/data/question_real_rekv_online_mod.json', 'w', encoding='utf-8') as f:
    json.dump(data, f, ensure_ascii=False, indent=2)
```
