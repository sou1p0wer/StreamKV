```python
import json

# 1. 读取json文件
json_path = '/home/cyl476530/ReKV/data/OVBench/ovbench_rekv.json'  # <-- 修改为你的json文件路径
with open(json_path, 'r', encoding='utf-8') as f:
    data = json.load(f)

# 2. 收集所有end_time连同video_id、video_path、question（如需要）
items = []
for video in data:
    video_id = video.get('video_id')
    video_path = video.get('video_path')
    for conv in video.get('conversations', []):
        if 'end_time' in conv:
            items.append({
                'video_id': video_id,
                'video_path': video_path,
                'end_time': conv['end_time'],
                'question': conv.get('question', '[NoQuestion]')
            })

# 3. 按end_time从大到小排序
items_sorted = sorted(items, key=lambda x: x['end_time'], reverse=True)

# 4. 打印结果
for i, item in enumerate(items_sorted, 1):
    print(f"{i}: video_id={item['video_id']}, end_time={item['end_time']}, video_path={item['video_path']}\n   question={item['question']}")

# 如果只想打印end_time，可这样：
# for i, item in enumerate(items_sorted, 1):
#     print(f"{i}: end_time={item['end_time']}")
```
