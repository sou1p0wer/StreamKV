```python
import json

# 设置文件名
input_file = '/home/cyl476530/ReKV/data/videomme/final.json'  # 原始JSON文件
output_file = '/home/cyl476530/ReKV/data/videomme/data.json'  # 替换后的JSON文件

# 路径替换
old_path = '/data/oss_bucket_0/video/Bench/Video-MME/videos/'
new_path = '/home/cyl476530/ReKV/data/videomme/videos/'

# 读取数据
with open(input_file, 'r', encoding='utf-8') as f:
    data = json.load(f)

# 替换 video_path
for item in data:
    if 'video_path' in item:
        item['video_path'] = item['video_path'].replace(old_path, new_path)

# 保存新文件
with open(output_file, 'w', encoding='utf-8') as f:
    json.dump(data, f, indent=2, ensure_ascii=False)

print("替换完成，新文件保存在:", output_file)
```
