```python
import json
import os

def time_str_to_sec(s):
    h, m, sec = map(int, s.split(':'))
    return h * 3600 + m * 60 + sec

def proc(streambench_json):
    all_data = []
    for item in streambench_json:
        video_path = item['video_path']
        video_id = os.path.splitext(os.path.basename(video_path))[0]
        conversations = []
        for q in item['questions']:
            st, ed = item['time'][1:-1].split(' - ')
            start_sec = time_str_to_sec(st)
            end_sec = time_str_to_sec(ed)
            conversation = dict(
                question = q['question'],
                answer = q.get('ground_truth_output', ''),  # 或其它字段
                start_time = start_sec,
                end_time = end_sec,
                ground_truth_time = time_str_to_sec(q['ground_truth_time_stamp']),
                ground_truth_output = q.get('ground_truth_output', ''),
                task_type = q.get('task_type', ''),
            )
            conversations.append(conversation)
        all_data.append(dict(
            video_path = video_path,
            video_id = video_id,
            conversations = conversations
        ))
    return all_data

# 用法
with open('/home/cyl476530/ReKV/data/StreamingBench/src/data/questions_proactive.json') as f:
    src = json.load(f)
dout = proc(src)
with open('/home/cyl476530/ReKV/data/StreamingBench/src/data/questions_proactive_rekv_online.json','w') as f:
    json.dump(dout, f, indent=2)
```
